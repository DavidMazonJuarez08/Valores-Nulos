```python
#Importamos las librerias pandas, numpy y matplotlib respectivamente
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Carga desde un archivo .xlsx sin indice
data = pd.read_excel('Principales_Canales.xlsx', index_col=0)
data.head()
```

```python
###ESTOS CÓDIGOS SON PARA ORDENAR Y LIMPIAR LOS DATOS DE "PISO"###
```

```python
#Visualizar número de filas
data.iloc[:, 0].head(50)
```

```python
#Seleccionar datos de Piso
piso = data.iloc[4:14, :]

piso
```

```python
#Usar la primera fila como nombres de columnas
piso.columns = piso.iloc[0]

#Eliminar la primera fila porque ya se utilizó como encabezado
piso = piso.iloc[1:]

piso
```

```python
#Transponer la tabla
piso = piso.T.reset_index()

#Cambiar nombre de la columna de los meses
piso = piso.rename(columns={"index": "Mes"})

piso
```

```python
#Cambiar nombre de la primera columna
piso = piso.rename(columns={piso.columns[0]: "Mes"})

#Quitar el nombre "Piso" que quedó arriba
piso.columns.name = None

piso.head()
```

```python
#Reemplazar guiones por valores nulos
piso = piso.replace("-", np.nan)
piso.head()
```

```python
#Contar valores nulos por columna
piso.isnull().sum()
```

```python
#Reemplazar valores nulos por 0
piso = piso.fillna(0)
```

```python
#Contar valores nulos por columna
piso.isnull().sum()
```
