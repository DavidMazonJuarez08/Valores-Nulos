```python
#Limpiar valores para convertir Potencial de compra a float
data1['Potencial de compra']=data1['Potencial de compra'].str.replace("Ventas ","0", regex=False)
data1['Potencial de compra']=data1['Potencial de compra'].str.replace("%","", regex=False)
data1['Potencial de compra']=data1['Potencial de compra'].str.replace(" ","", regex=False)
data1["Potencial de compra"]=data1["Potencial de compra"].str.replace('"',"", regex=False)
data1['Potencial de compra']=data1['Potencial de compra'].str.replace("&","", regex=False)
data1['Potencial de compra']=data1['Potencial de compra'].str.replace("'","", regex=False)
```

```python
#Corregir valores porcentuales
data1.loc[data1["Potencial de compra"]=="80","Potencial de compra"]="0.8"
```

```python
data1['Potencial de compra']=data1['Potencial de compra'].astype(float)
```

```python
#Corregir valor fuera de la escala
data1.loc[data1["Potencial de compra"]==8,"Potencial de compra"]=0.08
```

```python
data1["Potencial de compra"].unique()
```

```python
#Sustituir valores nulos con promedio o media
data1["Potencial de compra"]=data1["Potencial de compra"].fillna(round(data1["Potencial de compra"].median(),1))
```
